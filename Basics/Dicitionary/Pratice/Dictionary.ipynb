### Different datatypes in key and value

```python
d= {1:1, 2.5:2.5, True:True, 5+4j: 5+4j, 6: [5,7], 7:(5,7), 8:{5,7}, 9:{1:1, 2:2}}
```

```python
d # Everything can be accepted as Value, All singular data types as key
```

```python
# Trying Tuple as key
d = {(1,2):5}
```

```python
d
```

```python
#List as Key
d = {[1,2]: 5} # List is mutable hence it cannot be a key
```

```python
# Set as Key
d = {{1,2} : 5} # Set is also mutable, cannot be a key
```

```python
# Dictionary as a key
d = {{1:2}, 5} # Dictionary is also mutable, hence it also cannot be a key
```

### Dictionary Comprehensions

#### Iterable Pairs

```python
l1 = [(1, 'ONE'), (2, 'TWO'), (3, 'THREE'), (4, 'FOUR')]
```

```python
d1 = {x:y for x,y in l1}
```

```python
d1
```

```python
type(d1)
```

#### zip function in comprehension

```python
l2 = [1,2,3,4]
l3 = ['ONE', 'TWO','THREE', 'FOUR']
```

```python
d2 = {x:y for x,y in zip(l2, l3)}
```

```python
d2
```

#### enumeration function in comprehension

```python
d3 = {x:y for x,y in enumerate(l3, start=1)}
```

```python
d3
```
