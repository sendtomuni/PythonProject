### Exception Handling
To handle exception we have the following keyword: try, except, else, finally

```python
a = int(input('Enter value of A')) # Entering 10
b = int(input('Enter value of B')) # Entering 0

try:
    c = a // b
    print(f'a // b is {c}')
except:
    print(f'a // b is not possible as b is 0, Divide By 0 is not possible')

print('end')
```

### Using of Else in try -except

```python
L = [10,20,30,40,50]

try:
    index = int(input("Enter Index to get value")) # 10
    print(f'in array L at {index} value is {L[index]}')
except Exception as msg:
    print(msg)
else:
    print('Successfully value Fetched')
print('End of Program')
```

```python
L = [10,20,30,40,50]

try:
    index = int(input("Enter Index to get value")) # 2
    print(f'in array L at {index} value is {L[index]}')
except Exception as msg:
    print(msg)
else:
    print('Successfully value Fetched')
print('End of Program')
```

### Use of Finally

```python
L = [10,20,30,40,50]

try:
    index = int(input("Enter Index to get value")) # abc
    print(f'in array L at {index} value is {L[index]}')
except Exception as msg:
    print(msg)
else:
    print('Successfully value Fetched')
finally:
    print('End of Program')
```
