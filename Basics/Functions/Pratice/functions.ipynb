# How to Write a Function in Python
A function is a block of code that performs a specific task. In Python, you can define a function using the `def` keyword, followed by the function name and parentheses. Here's a simple example:

```python
def function_name(parameters):
    # Code block
    return result
```

## Formal Parameters
Formal parameters are the variables that are defined in the function's definition. They act as placeholders for the values that will be passed to the function when it is called. For example, in the function
```python
def add_numbers(a, b):
    return a + b
```
a and b are formal parameters. When you call the function, you provide actual values (arguments) for these parameters.

## Actual Parameters
Actual parameters, also known as arguments, are the values that you pass to a function when you call it. These values are used by the function to perform its task. For example, in the function call
```python
result = add_numbers(5, 3)
```
5 and 3 are actual parameters. The function will use these values to compute the result, which in this case will be 8.

# Supplying the Actual Parameters
When calling a function, you can supply the actual parameters in different ways:
1. **Positional Arguments**: The values are assigned to the parameters based on their position in the function call.
2. **Keyword Arguments**: You can specify the parameter names along with their values, allowing you to pass the arguments in any order.
3. **Default Arguments**: You can define default values for parameters in the function definition. If no value is provided for that parameter when calling the function, the default value will be used.
4. **Variable-length Arguments**: You can use `*args` for a variable number of positional arguments and `**kwargs` for a variable number of keyword arguments.

### Positional Arguments

````python
def volume(length, breadth, height):
    return length * breadth * height
````

````python
volume(5, 3, 2)  # Positional arguments: length=5, breadth=3, height=2
````

### Keyword Arguments

````python
volume(length=5, breadth=3, height=2)  # Keyword arguments: length=5, breadth=3, height=2
````

### Conditions in Positional Argument and Keyword Argument
When using keyword arguments, it should be aligned from right side. After a keyword argument, you cannot use positional arguments. For example:
```python
volume(length=5, 3, height=2)  # This will raise a SyntaxError
```

````python
volume(length=5, 3, height=2)  # This is not a valid as positional argument is used after a keyword argument.
````

Once a positional argument has been give, we cannot use the keyword of that position to pass the value. For example:
```python
volume(5, length=3, height=2) # This will raise a TypeError
```

````python
volume(5, length=3, height=2)  # This is not valid as the keyword argument 'length' is used after a positional argument of length.
````

While using the keyword argument, formal argument naming should be same as the keyword argument. For example:
```python
volume(length=5, breadth=3, height=2)  # This is valid
volume(length=5, width=3, height=2)  # This will raise a TypeError as 'width' is not a formal parameter of the function.
```

````python
volume(length=5, width=3, height=2)  # This is not valid as the keyword argument 'width' is not a formal parameter of the function.
````

```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"
```

````python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"
````

````python
greet("John", "Hi")  # Overrides the default greeting with "Hi"
````

### Default Argument behaviour
In Python, default parameter values are evaluated only once—at the exact moment the function is defined (when the module is loaded), not each time the function is called.

If you use a mutable object (like a list [], dictionary {}, or class instance) as a default argument, that same single object persists across every call to the function.

````python
def add_item_to_list(item, my_list=[]):
    my_list.append(item)
    return my_list
````

````python
add_item_to_list(1)  # Returns [1]
````

````python
add_item_to_list(2)  # Returns [1, 2] because the same
````

````python
add_item_to_list(3, [4])
````

````python
add_item_to_list(5)
````

Default argument values are not stored inside the function's local scope. Instead, they are stored directly on the function object itself as an attribute ```(__defaults__)```.

But we have option to avoid this behaviour by using None as the default value and then checking for it inside the function. For example:
```python
def add_item_to_list(item, my_list=None):
    if my_list is None:
        my_list = []
    my_list.append(item)
    return my_list
```

````python
def add_item_to_list(item, my_new_list=None):
    if my_new_list is None:
        my_new_list = []
    my_new_list.append(item)
    return my_new_list
````

````python
add_item_to_list(1)  # Returns [1]
````

````python
add_item_to_list(2)
````

````python
add_item_to_list(3, [4])  # Returns [4, 3]
````

````python
add_item_to_list(5)  # Returns [5]
````

Triaging the __defaults__ attribute of the function object:

**Note:** `__defaults__` only works on **Python-defined** functions, not on built-in methods (like `list.index`) which are implemented in C.
- For your own functions: use `function_name.__defaults__`
- For built-in methods: use `inspect.signature()` to see parameter info

````python
# For user-defined functions, __defaults__ shows the default values
print("volume defaults:", volume.__defaults__)  # (1,) because height=1
print("add_item_to_list defaults:", add_item_to_list.__defaults__)  # (None,) because my_new_list=None
````

````python
# For built-in methods like list.index(), use inspect.signature()
import inspect
print(inspect.signature(list.index))  # Shows (self, value, start=0, stop=9223372036854775807)
````

````python
# You can also inspect individual parameters and their defaults
import inspect
a = [10,20,30,40,50,20,30,40,50,20,30,40,50]
sig = inspect.signature(a.index)

for name, param in sig.parameters.items():
    if param.default is not inspect.Parameter.empty:
        print(f"{name} = {param.default}")
````

The reason inspect.signature shows 9223372036854775807 instead of len comes down to how built-in methods are stored in CPython:

🏷️ Class-Level Definition: The list.index method signature is created once for the general list class, before any specific list instance (with a specific length) is ever created.

📏 Dynamic Bounds: Because every list instance can have a different length, Python cannot store a single len value inside the static function signature metadata.

♾️ Universal Upper Bound: CPython sets the metadata default for stop to SYS_MAXSIZE (9223372036854775807). When the method actually executes on a specific list, it stops at whichever is smaller: the stop value or len(list).

### Positional-Only Parameters
In Python 3.8 and later, you can define parameters that can only be supplied positionally (not by keyword) using the `/` symbol in the function definition. For example:
```python
def func(a, b, /, c, d):
    return a + b + c + d
```
In this example, `a` and `b` are positional-only parameters, while `c and `d` can be supplied either positionally or by keyword. This is useful when you want to enforce a specific calling convention for certain parameters, especially in cases where the parameter names might be confusing or when you want to maintain backward compatibility with older code that relies on positional arguments.

````python
def func(a, b, /, c, d):
    print(f"a={a}, b={b}, c={c}, d={d}")
    return a + b + c + d
````

````python
func(1, 2, c=3, d=4)  # Valid: a=1, b=2, c=3, d=4
````

````python
func(1, 2, 3, 4)  # Valid: a=1, b=2, c=3, d=4
````

````python
func(a=1, b=2, c=3, d=4)  # Invalid: a and b are positional-only parameters
````

````python
func(1, 2, 3, d=4)  # Valid: a=1, b=2, c=3, d=4
````

````python
func(1, 2, c=3, 4)  # Invalid: positional
````

````python
func(1, b=2, 3, 4)  # Valid: a=1, b=2, c=3, d=4
````

In previous example even if I brake the rule of positional-only parameters, it will not raise any error to that, but it will raise a TypeError for positional argument follows keyword argument. This is because the positional-only parameters are defined before the `/` symbol, and they can only be supplied positionally. If you try to supply them as keyword arguments, Python will raise a TypeError indicating that the positional argument follows a keyword argument.

#### Rules for / in method argument:
1. **Positional-Only Parameters**: Any parameter defined before the `/` symbol is considered positional-only. You cannot use keyword arguments to supply values for these parameters.
2. **Positional or Keyword Parameters**: Parameters defined after the `/` symbol can be supplied either positionally or by keyword. You have the flexibility to choose how to pass values for these parameters.
3. **Order Matters**: The order of parameters in the function definition is important.
4. / is not a parameter: The `/` symbol itself is not a parameter; it is just a marker to indicate the end of positional-only parameters.
5. / cannot be placed at the beginning or end of the parameter list. It must be placed between parameters to separate positional-only parameters from those that can be supplied either positionally or by keyword.

### keyword-only Parameters
In Python, you can define parameters that can only be supplied by keyword using the `*` symbol in the function definition. For example:
```python
def func(a, b, *, c, d):
    return a + b + c + d
```
In this example, `c` and `d` are keyword-only parameters, while `a and `b` can be supplied either positionally or by keyword. This is useful when you want to enforce a specific calling convention for certain parameters, especially in cases where the parameter names might be confusing or when you want to maintain backward compatibility with older code that relies on positional arguments.

````python
def func(a, b, *, c, d):
    print(f"a={a}, b={b}, c={c}, d={d}")
    return a + b + c + d
````

````python
func(1, 2, c=3, d=4)  # Valid: a=1, b=2, c=3, d=4
````

````python
func(1, 2, 3, 4)  # Invalid: c and d are keyword-only parameters
````

````python
func(a=1, b=2, c=3, 4)  # Valid: a=1, b=2, c=3, d=4
````

````python
func(1, 2, c=3, 4)  # Invalid: positional argument follows keyword argument
````

#### Rules for * in method argument:
1. **Keyword-Only Parameters**: Any parameter defined after the `*` symbol is considered keyword-only. You cannot use positional arguments to supply values for these parameters.
2. **Positional or Keyword Parameters**: Parameters defined before the `*` symbol can be supplied either positionally or by keyword. You have the flexibility to choose how to pass values for these parameters.
3. **Order Matters**: The order of parameters in the function definition is important.
4. `*` is not a parameter: The `*` symbol itself is not a parameter; it is just a marker to indicate the beginning of keyword-only parameters.
5. `*` cannot be placed at the end of the parameter list. It must be placed between parameters to separate positional or keyword parameters from those that can only be supplied by keyword.
