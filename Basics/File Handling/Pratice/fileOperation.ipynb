### File Operations
There can be 2 options to read files:
###### 1. Sequential Access of Files
###### 2. Random Access of Files

In Sequential access of files, it will start reading from beginning of the file and get to end sequentially and read character by character.
In Random Access of files, system reads byte by byte and gives option to move the cursor to random places to read.

Let's say I have a file test.txt which has ABCDEFGHIJ content and here are the operations to read the files

```python
file = open('test.txt', 'rb')
# rb stands for read binary
```

```python
file.read(2)
```

output has been appended with b because system reads with bytes

```python
file.tell()
```

cursor is at position 2 now

```python
file.seek(4, 0)
```

first argument of seek how mnay bytes to move, second argument is positions.
0 -> stands for from Beginning
1 -> stands for from current position
2 -> stands for from End Of File

so in the previous line it says 0, so from beginning and position 4. So its in E now

```python
file.read(2)
```

and here it read 2 bytes again EF

```python
file.seek(2, 1)
```

I moved the cursor to 2 more position from 1(current position). Now curosr is at 8 (i.e I)

```python
file.read(1)
```

```python
file.tell()
```

```python
file.seek(-8, 1)
```

Now I am moving 8 position backward from current position. Cursor was at the 9(J). Now it moved to 1 (i.e B)

```python
file.read(3)
```

cursor is at 4 i.e E

```python
file.seek(-2,2)
```

I moved cursor 2 position backward form EOF(2), and now cursor move from 4(E) to 8(I)

```python
file.read(2)
```
