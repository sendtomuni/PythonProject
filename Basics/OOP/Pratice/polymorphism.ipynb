+ '+' is the same operator works differently for int and string. This is operator overloading

```python
def sum(sequence):
    sum = 0
    for i in sequence:
        sum += i
    return sum
```

```python
sum([2,4,6,8])
```

```python
sum((2,4,6,8))
```

```python
sum({2,4,6,8})
```

This is method overloading

```python
class Duck:
    def talk(self):
        print("Duck Talking")

    def walk(self):
        print("Duck Walking")
```

```python
class Dog:
    def talk(self):
        print("Dog Talking")

    def walk(self):
        print("Dog Walking")
```

```python
def person(pet):
    pet.talk()
    pet.walk()
```

```python
dog = Dog()
person(dog)
```

```python
duck = Duck()
person(duck)
```

This is called duck typing
