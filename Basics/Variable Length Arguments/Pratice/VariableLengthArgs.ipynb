# Variable Length Arguments
There are two types of variable length arguments in Python:
1. *args (Non-keyword variable length arguments)
2. **kwargs (Keyword variable length arguments)

```python
# Example of *args
def my_function(*args):
    for arg in args:
        print(arg)
```

```python
my_function(1, 2, 3, 4, 5)
```

```python
my_function("Hello", "World", "!")
```

```python
def function_with_args(*args):
    print(args, type(args), len(args))
```

```python
function_with_args("apple", "banana", "cherry")
```

```python
my_function(name="Alice", age=30, city="New York")
```

#### Ways of defining a function with variable length arguments

```python
def fun_var_args(a, b, *args):
    print("a:", a)
    print("b:", b)
    print("args:", args)
```

```python
fun_var_args(1, 2, 3, 4, 5)
```

```python
fun_var_args((1,2), (3,4), ((5,6), (7,8)))
```

```python
fun_var_args((1,2), (3,4), (5,6), (7,8))
```

```python
fun_var_args(a=10, b=20, args = (30, 40, 50))
```

```python
fun_var_args(a=10, b=20, 30, 40, 50)
```

```python
def fun_var_args2(*args, a, b):
    print("a:", a)
    print("b:", b)
    print("args:", args)
```

```python
fun_var_args2(10, 20, 30, 40, a=50, b=60)
```

```python
fun_var_args2(10,20,30,40,50)
```

```python
fun_var_args2((10,20,30),40,50)
```

```python
def fun_var_args3(a, b, *args, c=100, d=200):
    print("a:", a)
    print("b:", b)
    print("args:", args)
    print("c:", c)
    print("d:", d)
```

```python
fun_var_args3(1, 2, 3, 4, 5, c=10, d=20)
```

```python
fun_var_args3(1, 2, 3, 4, 5)
```

```python
fun_var_args3(1, 2, c=10, d=20)
```

```python
fun_var_args3(1)
```

```python
L1 = [1, 2, 3, 4, 5]
```

```python
function_with_args(L1)
```

```python
function_with_args(*L1)
```

```python
def fun_keyword_args(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
fun_keyword_args(name="Alice", age=30, city="New York")
```

```python
def fun_keyword_args2(a, b, **kwargs):
    print("a:", a)
    print("b:", b)
    print("kwargs:", kwargs)
```

```python
fun_keyword_args2(1, 2, name="Alice", age=30, city="New York")
```

```python
fun_keyword_args2(a=1, b=2, name="Alice", age=30)
```

```python
def fun_keyword_args3(**kwargs, a, b):
    print("a:", a)
    print("b:", b)
    print("kwargs:", kwargs)
```

```python
def fun_keyword_args4(a, b, *args, c=100, d=200, **kwargs):
    print("a:", a)
    print("b:", b)
    print("args:", args)
    print("c:", c)
    print("d:", d)
    print("kwargs:", kwargs)
```

```python
fun_keyword_args4(1, 2, 3, 4, 5, c=10, d=20, name="Alice", age=30)
```

```python
fun_keyword_args4(1, 2, name="Alice", age=30)
```

```python
fun_keyword_args4(1, 2, c=10, d=20, name="Alice", age=30)
```

```python
def fun_keyword_args5(**kwargs, *args):
    print("args:", args)
    print("kwargs:", kwargs)
```

# Note: The order of parameters in a function definition is important. You cannot have *args after **kwargs or vice versa. The correct order is:
# 1. Positional arguments
# 2. *args
# 3. Keyword arguments

### Multiple Returns

Python functions can return multiple values using tuples. When you return multiple values, they are packed into a tuple and can be unpacked when calling the function.

```python
def fun_sum_product(a, b):
    return a + b, a * b
```

```python
print(fun_sum_product(3, 4))
```

Returning multiple related values from a single fucntion groups logic together, making code more maintainable and reducing duplication. This encapsulation of logic is a key principle of software design, promoting modularity and reusability.
